# Step 1:- Importing Required Libraries & Loading the Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('Titanic-Dataset.csv')

# Display first few rows
df.head()
```

# Step 2: Basic Information About the Dataset

```python
# Dataset overview
df.info()

# Summary statistics
df.describe()

# Count of unique values in each column
df.nunique()
```

# Step 3: Checking for Missing Values

```python
df.isnull().sum()
```

# Step 4: Distribution of Categorical Columns

```python
# Value counts for categorical columns
print(df['Sex'].value_counts())
print(df['Pclass'].value_counts())
print(df['Embarked'].value_counts())
print(df['Survived'].value_counts())
```

# Step 5: Pairplot to Understand Pairwise Relationships

```python
# Pairplot (only numerical columns)
sns.pairplot(df, hue='Survived')
plt.show()
```

# Step 6: Heatmap to Visualize Correlation

```python
# Select only numeric columns
numeric_df = df.select_dtypes(include='number')

# Correlation heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap (Numeric Features)")
plt.show()
```

# Step 7: Histograms for Distribution

```python
# Plot histograms for all numeric features
df.hist(figsize=(12, 10), bins=30, edgecolor='black')
plt.suptitle("Histograms of Numeric Features")
plt.tight_layout()
plt.show()
```

# Step 8: Boxplots to Check for Outliers

```python
# Boxplots for selected columns
plt.figure(figsize=(12, 6))
sns.boxplot(x='Pclass', y='Age', data=df)
plt.title("Age distribution by Passenger Class")
plt.show()

sns.boxplot(x='Survived', y='Fare', data=df)
plt.title("Fare distribution by Survival")
plt.show()
```

# Step 9: Scatterplots to Explore Relationships

```python
# Age vs Fare with hue as Survived
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived')
plt.title("Age vs Fare Colored by Survival")
plt.show()
```
